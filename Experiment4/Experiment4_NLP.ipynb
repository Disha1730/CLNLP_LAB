```python
import nltk
from collections import Counter

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')

text = """
Natural Language Processing is a field of Artificial Intelligence.
Google developed many technologies using Machine Learning.
John works at Microsoft in New York.
"""

# ---------------- TERM FREQUENCY ----------------

# Tokenize text
words = nltk.word_tokenize(text.lower())

# Keep only words
words = [word for word in words if word.isalpha()]

# Calculate frequency
frequency = Counter(words)

print("Term Frequency:")
print("-------------------------")

for word, count in frequency.items():
    print(word, ":", count)


# ---------------- TOP 10 ----------------

print("\nTop 10 Most Frequent Terms:")
print("-------------------------")

top_10 = frequency.most_common(10)

for word, count in top_10:
    print(word, ":", count)


# ---------------- NER ----------------

# Tokenize original text
tokens = nltk.word_tokenize(text)

# POS tagging
tagged_words = nltk.pos_tag(tokens)

# Named Entity Recognition
entities = nltk.ne_chunk(tagged_words)

print("\nNamed Entities:")
print("-------------------------")

for entity in entities:

    if hasattr(entity, 'label'):

        name = " ".join(
            word for word, tag in entity.leaves()
        )

        print(name, ":", entity.label())
```

Experiment 4.2 — Term Frequency WITHOUT Toolkit

```python
import re
import csv
from collections import Counter

text = """
Natural Language Processing is a field of Artificial Intelligence.
Google developed many technologies using Machine Learning.
John works at Microsoft in New York.
"""

# Convert text to lowercase
text = text.lower()

# Extract words
words = re.findall(r'\b[a-z]+\b', text)

# Calculate frequency
frequency = Counter(words)

# ---------------- SAVE TO CSV ----------------

with open("term_frequency.csv", "w", newline="") as file:

    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Data
    for word, count in frequency.items():
        writer.writerow([word, count])

print("Term frequency saved to term_frequency.csv")

# ---------------- DISPLAY FREQUENCY ----------------

print("\nTerm Frequency:")

for word, count in frequency.items():
    print(word, ":", count)

# ---------------- TOP 10 ----------------

print("\nTop 10 Most Frequent Terms:")

top_10 = frequency.most_common(10)

for word, count in top_10:
    print(word, ":", count)
```

Experiment 4.3 — TF, DF, IDF and TF-IDF WITHOUT NLP Toolkit

```python
import re
import math
from collections import Counter

# ---------------- DOCUMENTS ----------------

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# ---------------- PREPROCESSING ----------------

processed_documents = []

for document in documents:

    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    document = re.sub(r'[^a-z\s]', '', document)

    # Tokenize
    words = document.split()

    processed_documents.append(words)


# Display tokens
print("TOKENS")
print("================================")

for i, words in enumerate(processed_documents):
    print("Document", i + 1, ":", words)


# ---------------- TF ----------------

print("\n\nTERM FREQUENCY (TF)")
print("================================")

tf_documents = []

for words in processed_documents:

    total_words = len(words)

    word_count = Counter(words)

    tf = {}

    for word, count in word_count.items():
        tf[word] = count / total_words

    tf_documents.append(tf)

    print("\nDocument:", len(tf_documents))

    for word, value in tf.items():
        print(word, ":", round(value, 4))


# ---------------- DF ----------------

print("\n\nDOCUMENT FREQUENCY (DF)")
print("================================")

df = {}

# Get all unique terms
all_terms = set()

for words in processed_documents:
    all_terms.update(words)

# Calculate DF
for term in sorted(all_terms):

    count = 0

    for words in processed_documents:

        if term in words:
            count += 1

    df[term] = count

for term, value in df.items():
    print(term, ":", value)


# ---------------- IDF ----------------

print("\n\nINVERSE DOCUMENT FREQUENCY (IDF)")
print("================================")

N = len(documents)

idf = {}

for term in sorted(df):

    idf[term] = math.log(N / df[term])

    print(
        term,
        ":",
        round(idf[term], 4)
    )


# ---------------- TF-IDF ----------------

print("\n\nTF-IDF")
print("================================")

tfidf_documents = []

for i in range(N):

    tfidf = {}

    for term in tf_documents[i]:

        tf_value = tf_documents[i][term]

        idf_value = idf[term]

        tfidf_value = tf_value * idf_value

        tfidf[term] = tfidf_value

    tfidf_documents.append(tfidf)

    print("\nDocument", i + 1)

    print("Term\t\tTF\t\tDF\t\tIDF\t\tTF-IDF")

    for term in sorted(tfidf):

        print(
            term,
            "\t\t",
            round(tf_documents[i][term], 4),
            "\t\t",
            df[term],
            "\t\t",
            round(idf[term], 4),
            "\t\t",
            round(tfidf[term], 4)
        )


# ---------------- TOP 10 TF-IDF ----------------

print("\n\nTOP 10 TERMS WITH HIGHEST TF-IDF")
print("================================")

for i in range(N):

    print("\nDocument", i + 1)

    sorted_terms = sorted(
        tfidf_documents[i].items(),
        key=lambda x: x[1],
        reverse=True
    )

    top_10 = sorted_terms[:10]

    for term, score in top_10:

        print(
            term,
            ":",
            round(score, 4)
        )
```
