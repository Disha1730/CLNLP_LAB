Experiment-1: Data Analysis


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/drive/MyDrive/employee_information_100.csv")
```

```python
df.head()
```

Experiment 1.1-Average Salary of Each Department (Bar Chart)




```python
avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)

plt.figure(figsize=(8,5))
avg_salary.plot(kind="bar")

plt.title("Average Salary of Each Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.show()
```

Experiment 1.2
Number of Employees in Each Department

```python
department_count = df["Department"].value_counts()

print(department_count)

plt.figure(figsize=(8,5))
department_count.plot(kind="bar")

plt.title("Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

Experiment 1.3
Percentage of Male and Female Employees (Pie Chart)


```python
gender_count = df["Gender"].value_counts()

print(gender_count)

plt.figure(figsize=(6,6))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

Experiment 1.4
Salary Distribution (Histogram)


```python
plt.figure(figsize=(8,5))

plt.hist(df["Salary"], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

Experiment 1.5
Experience vs Salary (Scatter Plot)

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

departments = df["Department"].unique()

colors = ["red", "blue", "green", "orange", "purple", "brown", "pink"]

for i, dept in enumerate(departments):
    data = df[df["Department"] == dept]

    plt.scatter(
        data["Experience_Years"],
        data["Salary"],
        color=colors[i % len(colors)],
        label=dept,
        s=70
    )

plt.title("Experience vs Salary by Department")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.legend(title="Department")
plt.grid(True)

plt.show()
```

Experiment 1.6
Top 10 Highest Paid Employees

```python
top10 = df.sort_values(by="Salary", ascending=False).head(10)

print(top10)
```

Experiment 1.7
Highest Salary in Every Department


```python
highest_salary = df.groupby("Department")["Salary"].max()

print(highest_salary)
```

Experiment 1.8

```python
average_salary = df["Salary"].mean()

print("Average Salary =", average_salary)

result = df[df["Salary"] > average_salary]

print(result)
```
